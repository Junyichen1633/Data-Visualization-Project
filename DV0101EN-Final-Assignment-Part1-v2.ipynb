<br>2. Recession: A binary variable indicating recession perion; 1 means it was recession, 0 means it was normal.
<br>3. Automobile_Sales: The number of vehicles sold during the period.
<br>4. GDP: The per capita GDP value in USD.
<br>5. Unemployment_Rate: The monthly unemployment rate.
<br>6. Consumer_Confidence: A synthetic index representing consumer confidence, which can impact consumer spending and automobile purchases.
<br>7. Seasonality_Weight: The weight representing the seasonality effect on automobile sales during the period.
<br>8. Price: The average vehicle price during the period.
<br>9. Advertising_Expenditure: The advertising expenditure of the company.
<br>10.Vehicle_Type: The type of vehicles sold; Supperminicar, Smallfamiliycar,                 Mediumfamilycar, Executivecar, Sports.
<br>11.Competition: The measure of competition in the market, such as the number of competitors or market share of major manufacturers.
<br>12.Month: Month of the observation extracted from Date..
<br>13.Year: Year of the observation extracted from Date.
<br>

### TASK 1.1: Develop a *Line chart* using the functionality of pandas to show how automobile sales fluctuate from year to year


<details><summary>Click here for a hint</summary>

<p>

````python
import pandas as pd
import matplotlib.pyplot as plt

# Grouping the data by 'Year' and calculating the mean of 'Automobile_Sales'
sales_by_year = df.groupby('Year')['Automobile_Sales'].mean()

# Plotting the data
plt.figure(figsize=(12, 6))
sales_by_year.plot(kind='line', marker='o')

# Customizing the plot
plt.title('Automobile Sales during Recession')
plt.xlabel('Year')
plt.ylabel('Average Automobile Sales')
plt.grid(True)
plt.xticks(rotation=45)

# Annotating at least two recession years
plt.axvline(x=1980, color='red', linestyle='--', label='Recession 1980')
plt.axvline(x=2008, color='orange', linestyle='--', label='Recession 2008')

plt.legend()
plt.tight_layout()
plt.savefig("Line_Plot_1.png")
plt.show()
````

####  TASK 1.2: Plot different lines for categories of vehicle type and analyse the trend to answer the question Is there a noticeable difference in sales trends between different vehicle types during recession periods?

````python
df_grouped = df.groupby(['Year', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()
import matplotlib.pyplot as plt

# Set up the figure
plt.figure(figsize=(12, 6))

# Loop through each unique vehicle type and plot
for vehicle in df_grouped['Vehicle_Type'].unique():
    subset = df_grouped[df_grouped['Vehicle_Type'] == vehicle]
    plt.plot(subset['Year'], subset['Automobile_Sales'], label=vehicle)

# Add plot title and labels
plt.title('Automobile Sales Trends by Vehicle Type')
plt.xlabel('Year')
plt.ylabel('Average Automobile Sales')
plt.xticks(rotation=45)
plt.grid(True)
plt.legend()

# Save and show plot
plt.tight_layout()
plt.savefig('Line_Plot_2.png')
plt.show()
````

### TASK 1.4: Use sub plotting to compare the variations in GDP during recession and non-recession period by developing line plots for each period.

### TASK 1.5: Develop a Bubble plot for displaying the impact of seasonality on Automobile Sales.
<br>How has seasonality impacted the sales, in which months the sales were high or low? Check it for non-recession years to understand the trend

##### Develop a Bubble plot for displaying Automobile Sales for every month and use Seasonality Weight for representing the size of each bubble<br>
Title this plot as 'Seasonality impact on Automobile Sales'


<details><summary>Click here for a hint</summary>

<p>
     You can create Bubble Chart by calling the scatter()
        <br>Pass the 'Month' and 'Automobile_Sales' to the functions as x and y and then use Seasonality weight for size parameter</p>
</details>


````python
df['Month_Num'] = pd.to_datetime(df['Date']).dt.month
df_bubble = df.groupby('Month_Num')[['Automobile_Sales', 'Seasonality_Weight']].mean().reset_index()
plt.figure(figsize=(12, 6))

# Use scatter plot where size is Seasonality_Weight
plt.scatter(
    df_bubble['Month_Num'],
    df_bubble['Automobile_Sales'],
    s=df_bubble['Seasonality_Weight'] * 500,  # scale bubble size
    alpha=0.6,
    color='skyblue',
    edgecolors='black'
)

# Customize
plt.title('Seasonality Impact on Automobile Sales')
plt.xlabel('Month')
plt.ylabel('Average Automobile Sales')
plt.xticks(ticks=range(1, 13))
plt.grid(True)

# Save and show
plt.tight_layout()
plt.savefig('Bubble.png')
plt.show()
````
